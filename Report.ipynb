## Formula 1


$$
\mathfrak{U} \doteq a \texttt{;} b \land \neg \exists p, s : b = p \texttt{;} s
$$

<div align="center">

```
$U = a ";" b && ¬∃ p ∃ s b = p ";" s
```

</div>

> "Find the last string that appears after a semicolon"


````python
formula = '$U = a ";" b && ¬∃ p ∃ s b = p ";" s'

words = ['aaabbbb;cccc;ddd;zzz', 'aaaaa;bb', ';', 'hello there',
         "actually quite a lot to say here;lskdjf slkdfjlskdfjhlskjdhflskjdhflksjdhflksjdhf;sdfkjhsdlfkjhsdlfkjhsldkfjhsldkjfhslkdjfhlskdjfhlskdjfhlskjdfhslkdjfhslkdjfhsldkjfhsdfs ;sjklj"]
````

````python
results = test_runner.benchmark_branches(
    cases=fc_test_cases(formula, *words),
    branches=["groundup-lcp", "ft-nuc-nohash-lcp-enum"],
    num_trials=3
)
````

````python
plot_results(PerfEvent.CACHE_MISSES, results,  perc=True)
````

## Formula 2

$$
l \doteq \texttt{;} c \texttt{;} \land c = a \texttt{,} b \land \neg (\exists p, s : c = p \texttt{;}s) \land \neg (\exists p, s : a = p \texttt{,}s)
$$

<div align="center">

```
l = ";"c";" && c = a "," b && ¬(∃ p ∃ s  c = p ";"  s) && ¬(∃ p ∃ s  a = p "," s)
```

</div>

> A kind of CSV-inspired parser, where each row is surrounded by semi-colons,
> and each of the two columns are delimited by commas.
>
> Each line $l$ has line-content $c$ that does not contain a `;`, and contains
> values $a$ and $b$ that do not contain a `,`


````python
formula = 'l=";"c";" && c=a "," b && ¬(exists p exists s  c = p ";"  s) && ¬(exists p exists s  a = p "," s)'

words = [';aa,cc;11,22;something here,more content;and more,here;']
results = test_runner.benchmark_branches(
    cases=fc_test_cases(formula, *words),
    branches=["groundup-lcp", "groundup-lcp-join"],
    num_trials=3
)
````

````python
plot_results(PerfEvent.DURATION_TIME, results)
````
