## Comparison of Hashing algorithms

## Simulated comparison of substring operations


```python
strings_run_command = ["./target/release/StringBenchmarker", "$WORD$"]
strings_test_runner = marker.Benchmarker(
    "/home/ahutton/dev/uni/Part D Project/StringBenchmarker", run_command=strings_run_command)
```

```python
strings_test_runner.repo.list_branches()
```

```python
odyssey = open("./test cases/texts/odyssey trimmed.txt").read()
xmas = open("./test cases/texts/short-xmas.txt").read()
ying_han = open("./test cases/texts/ying-han-trimmed.txt").read()

case_names = ["English", "Greek", "Chinese"]
strings_bench_results = strings_test_runner.benchmark_branches(
    [
        {"$WORD$": xmas[:1000]},
        {"$WORD$": odyssey[:1000]},
        {"$WORD$": ying_han[:1000]},
    ],
    num_trials=3,
    print_output=False,
)
```

```python
from datetime import datetime
from time import time
import json

timestamp = datetime.fromtimestamp(time()).strftime("%Y-%m-%d %H-%M-%S")

with open(f"./out/synthetic-substrings-for-report-{timestamp}.json", "w") as f:
    json.dump(strings_bench_results, f, indent=4)
```
