```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_formula_performance(formula, bench_results):
    branch_names = bench_results.keys()

    # list of results for formula across test words across trials, by branch
    branches_word_results = [
        bench_results[branch][formula] for branch in branch_names
    ]
    print(branches_word_results)

    # Average result across trials
    branches_word_results = [
        {word: np.mean(vals) for (word, vals) in branch.items()} for branch in branches_word_results
    ]

    words = branches_word_results[0]
    x = np.arange(len(branch_names))

    times_by_word = {word: [] for word in branches_word_results[0].keys()}
    for i, branch in enumerate(branches_word_results):
        for (word, value) in branch.items():
            times_by_word[word].append(value)
    # print(times_by_word)

    # plt assumes $ means LaTeX, so escape
    plt.title(formula.replace("$", "\\$"))
    for i, (word, times) in enumerate(times_by_word.items()):
        # print(times)
        plt.bar(x + (i-1.5)/(len(words)+1), times,
                width=1/(len(words)+1), label=word)
    plt.xticks(x, list(branch_names))
    plt.ylabel("Mean runtime for formula on word (s)")
    # plt.ylim(0, 10)
    plt.legend()


plot_formula_performance('Z = X "ab"Y && Z = Y "ab"X', bench_results)
```
